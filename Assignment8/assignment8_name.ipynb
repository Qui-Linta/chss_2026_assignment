## 課題1

以下の図を再現してください

- 年齢の分布を、異なる階級幅を使って、4つのサブプロットをわけて描画する
   - 階級幅を適当で決めてよい
- 各サブプロットには、
    - x軸ラベル、y軸ラベル、サブプロット番号(a-d)を指定する
    - 水平方向のグリッド線を追加する
    - ヒストグラムの色を`lightblue`に変更する

![](./Figure1.png)

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2)

bins = 0
for i in range(2):
    for j in range(2):
        bins += 1
        axs[i, j].hist(df[(50 * (bins - 1) < df['fare']) & (df['fare'] < 50 * bins)]['age'], color='lightblue', density=True)
        axs[i, j].set_title(f'fare:{50 * (bins - 1)} ~ {50 * bins}')
        axs[i, j].set_xlabel('age')
        axs[i, j].set_ylabel('density')

plt.tight_layout()
plt.show()
```

## 課題2

以下の図を再現してください

- 男性乗客と女性乗客の年齢の分布をそれぞれ描画する
- 全乗客の年齢分布を示す灰色ヒストグラムも描画する
- 図のタイトル、x軸ラベル、y軸ラベル、凡例を設定する

```python
import pandas as pd
import matplotlib.pyplot as plt


df=pd.read_csv("https://raw.githubusercontent.com/lvzeyu/css_tohoku/master/css_tohoku/draft/Data/titanic.csv")

fig, axs = plt.subplots(2)

axs[0].hist(df['age'], color='gray', bins=40)
axs[0].hist(df[(df['sex'] == 'male')]['age'], color='blue', bins=40)

axs[0].legend(['すべての乗客', '男性乗客'])
axs[0].set_title('男性乗客')
axs[0].set_xlabel('乗客の年齢(歳)')
axs[0].set_ylabel('人数')

axs[1].hist(df['age'], color='gray', bins=40)
axs[1].hist(df[(df['sex'] == 'female')]['age'], color='red', bins=40)

axs[1].legend(['すべての乗客', '女性乗客'])
axs[1].set_title('女性乗客')
axs[1].set_xlabel('乗客の年齢(歳)')
axs[1].set_ylabel('人数')

plt.tight_layout()
plt.show()
```
