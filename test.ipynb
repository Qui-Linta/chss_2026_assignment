```python
import matplotlib.pyplot as plt
import scienceplots
from matplotlib import font_manager


#plt.style.use('classic')
plt.style.use(['science','no-latex'])

# Path to your TTF file
ttf_path = "C:/Users/femsq/OneDrive/ドキュメント/Classroom/Noto_Sans_JP/Noto_Sans_JP/NotoSansJP-VariableFont_wght.ttf"

# Register the font
font_manager.fontManager.addfont(ttf_path)
custom_font = font_manager.FontProperties(fname=ttf_path)

# インストールされているフォント名を取得して表示
fonts = set([f.name for f in font_manager.fontManager.ttflist])

# 日本語フォントによく含まれるキーワードで絞り込み
japanese_fonts = [f for f in fonts if any(kw in f.lower() for kw in ['jp', 'gothic', 'mincho', 'meiryo', 'yu', 'ms'])]

for font in sorted(japanese_fonts):
    print(font)

# Set the custom font as default
plt.rcParams['font.family'] = custom_font.get_name()
plt.rcParams['font.family'] = 'HGMinchoE'

city=['東京', '横浜', '大阪', '名古屋', '札幌', '神戸', '京都', '福岡', '広島', '仙台']
population= [13929286, 3723392, 2691004, 2390411, 1952356, 1538316, 1474570, 1532591, 1196564, 1098330]
plt.figure(figsize=(6,3))
plt.bar(city,population)
plt.show()
```
